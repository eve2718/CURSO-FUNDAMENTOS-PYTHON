```python
def calculadora():
    while True:
        try:
            num1 = float(input("Ingrese el primer número (o 'q' para salir): "))
            num2 = float(input("Ingrese el segundo número: "))
            operador = input("Ingrese el operador (+, -, *, /, **, //): ")

            operaciones = {
                '+': num1 + num2,
                '-': num1 - num2,
                '*': num1 * num2,
                '/': num1 / num2 if num2 != 0 else "Error: División por cero.",
                '**': num1 ** num2,
                '//': num1 // num2 if num2 != 0 else "Error: División por cero."
            }

            if operador in operaciones:
                resultado = operaciones[operador]
                print(f"Resultado: {num1} {operador} {num2} = {resultado}")
            else:
                print("Operador no válido.")
        except ValueError:
            print("Error: Entrada no válida. Asegúrese de ingresar números.")
        except Exception as e:
            print(f"Error: {e}")

calculadora()
```

```python
def administrador_inventarios():
    inventario = {'manzanas': 50, 'naranjas': 30, 'peras': 20}

    def mostrar_inventario():
        print("Inventario actual:")
        for producto, cantidad in inventario.items():
            print(f"{producto}: {cantidad}")

    def agregar_producto():
        producto = input("Ingrese el nombre del producto: ")
        cantidad = int(input("Ingrese la cantidad: "))
        inventario[producto] = inventario.get(producto, 0) + cantidad
        print(f"Producto '{producto}' agregado con {cantidad} unidades.")

    def actualizar_producto():
        producto = input("Ingrese el nombre del producto a actualizar: ")
        if producto in inventario:
            cantidad = int(input("Ingrese la nueva cantidad: "))
            inventario[producto] = cantidad
            print(f"Producto '{producto}' actualizado a {cantidad}.")
        else:
            print(f"Error: El producto '{producto}' no existe.")

    def eliminar_producto():
        producto = input("Ingrese el nombre del producto a eliminar: ")
        if producto in inventario:
            del inventario[producto]
            print(f"Producto '{producto}' eliminado.")
        else:
            print(f"Error: El producto '{producto}' no existe.")

    while True:
        print("\nOpciones: 1. Mostrar inventario 2. Agregar producto 3. Actualizar producto 4. Eliminar producto 5. Salir")
        opcion = input("Seleccione una opción: ")
        if opcion == '1':
            mostrar_inventario()
        elif opcion == '2':
            agregar_producto()
        elif opcion == '3':
            actualizar_producto()
        elif opcion == '4':
            eliminar_producto()
        elif opcion == '5':
            break
        else:
            print("Opción no válida.")

administrador_inventarios()
```

```python
import re

def analisis_texto():
    texto = input("Ingrese un texto: ")
    if not texto.strip():
        print("Error: No se ingresó texto.")
        return

    palabras = re.findall(r'\b\w+\b', texto.lower())
    frecuencia = {}

    for palabra in palabras:
        frecuencia[palabra] = frecuencia.get(palabra, 0) + 1

    print("Frecuencia de palabras:")
    for palabra, count in sorted(frecuencia.items()):
        print(f"{palabra}: {count}")

analisis_texto()
```

```python
def conversor_unidades():
    opciones = {
        '1': ("Centímetros a pulgadas", lambda cm: cm / 2.54),
        '2': ("Kilómetros a millas", lambda km: km * 0.621371)
    }

    print("Seleccione la conversión:")
    for key, (descripcion, _) in opciones.items():
        print(f"{key}. {descripcion}")

    opcion = input("Opción: ")
    if opcion in opciones:
        cantidad = float(input(f"Ingrese la cantidad en {opciones[opcion][0].split()[0].lower()}: "))
        resultado = opciones[opcion][1](cantidad)
        print(f"Resultado: {cantidad} = {resultado:.2f} {opciones[opcion][0].split()[-1]}")
    else:
        print("Opción no válida.")

conversor_unidades()
```

```python
import random

def adivina_el_numero():
    numero_secreto = random.randint(1, 50)
    intentos = 0

    print("¡Bienvenido al juego 'Adivina el Número'!")
    print("Estoy pensando en un número entre 1 y 50.")

    while True:
        intento = input(f"Intento {intentos + 1} - Ingresa tu número (o 'q' para salir): ")
        if intento.lower() == 'q':
            print("Juego terminado.")
            break
        try:
            intento = int(intento)
            intentos += 1

            if intento < numero_secreto:
                print("El número secreto es mayor.")
            elif intento > numero_secreto:
                print("El número secreto es menor.")
            else:
                print(f"¡Felicidades! Has adivinado el número secreto: {numero_secreto} en {intentos} intentos.")
                break
        except ValueError:
            print("Error: Por favor, ingresa un número válido.")

adivina_el_numero()
```

```python
from datetime import datetime

def manejo_fechas():
    while True:
        fecha_str = input("Ingrese una fecha (DD/MM/AAAA) o 'q' para salir: ")
        if fecha_str.lower() == 'q':
            print("Saliendo del programa.")
            break
        try:
            fecha = datetime.strptime(fecha_str, "%d/%m/%Y")
            hoy = datetime.now()
            diferencia = (fecha - hoy).days
            print(f"Faltan {diferencia} días para el {fecha_str}.")
            break
        except ValueError:
            print("Error: Formato de fecha no válido. Intente nuevamente.")

manejo_fechas()
```

```python
def registro_estudiantes():
    registro = {}

    def mostrar_menu():
        print("\nMenú:")
        print("1. Agregar estudiante")
        print("2. Actualizar calificación")
        print("3. Eliminar estudiante")
        print("4. Listar estudiantes")
        print("5. Salir")

    while True:
        mostrar_menu()
        opcion = input("Opción: ")

        if opcion == '1':
            nombre = input("Ingrese el nombre del estudiante: ")
            calificacion = float(input("Ingrese la calificación: "))
            registro[nombre] = calificacion
            print(f"Estudiante '{nombre}' agregado con calificación {calificacion}.")
        elif opcion == '2':
            nombre = input("Ingrese el nombre del estudiante: ")
            if nombre in registro:
                calificacion = float(input("Ingrese la nueva calificación: "))
                registro[nombre] = calificacion
                print(f"Estudiante actualizado: {nombre} - {calificacion}")
            else:
                print(f"Error: El estudiante '{nombre}' no existe.")
        elif opcion == '3':
            nombre = input("Ingrese el nombre del estudiante a eliminar: ")
            if nombre in registro:
                del registro[nombre]
                print(f"Estudiante '{nombre}' eliminado.")
            else:
                print(f"Error: El estudiante '{nombre}' no existe.")
        elif opcion == '4':
            print("Registro de estudiantes:", registro)
        elif opcion == '5':
            break
        else:
            print("Opción no válida.")

registro_estudiantes()
```

```python
import random
import string

def generador_contrasenas():
    while True:
        longitud = input("Ingrese la longitud deseada para la contraseña (o 'q' para salir): ")
        if longitud.lower() == 'q':
            print("Saliendo del generador de contraseñas.")
            break
        try:
            longitud = int(longitud)
            if longitud <= 0:
                print("Error: La longitud debe ser un número entero positivo.")
                continue

            caracteres = string.ascii_letters + string.digits + string.punctuation
            contrasena = ''.join(random.choice(caracteres) for _ in range(longitud))
            print(f"Contraseña generada: {contrasena}")
            break
        except ValueError:
            print("Error: Entrada no válida. Asegúrese de ingresar un número entero.")

generador_contrasenas()
```

```python
import os

def buscador_archivos():
    while True:
        directorio = input("Ingrese la ruta del directorio (o 'q' para salir): ")
        if directorio.lower() == 'q':
            print("Saliendo del buscador de archivos.")
            break
        subcadena = input("Ingrese la subcadena a buscar: ")

        try:
            archivos_encontrados = [archivo for archivo in os.listdir(directorio) if subcadena in archivo]
            if archivos_encontrados:
                print(f"Archivos encontrados que contienen '{subcadena}':")
                for archivo in archivos_encontrados:
                    print(archivo)
            else:
                print(f"No se encontraron archivos que contengan '{subcadena}'.")
        except FileNotFoundError:
            print("Error: El directorio no existe.")
        except PermissionError:
            print("Error: Permisos insuficientes para acceder al directorio.")

buscador_archivos()
```
